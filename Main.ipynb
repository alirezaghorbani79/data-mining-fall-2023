```python
import numpy as np
import pandas as pd
```

```python
DATASET_PATH = './Dataset.csv'
df = pd.read_csv(DATASET_PATH)

df_head = df.head()
df_info = df.info()
df_description = df.describe()

print(df_head, df_info, df_description)
```

```python
zero_variance_cols = [col for col in df.columns if df[col].nunique() == 1]

#Drop zaro variance colomns
df_cleaned = df.drop(columns=zero_variance_cols)

df_cleaned.info()
```

```python
low_variability_cols = {col: df[col].nunique() for col in df.columns if df[col].nunique() < 10}

print("Columns with Very Few Unique Values:")
for col, unique_count in low_variability_cols.items():
    print(f"{col}: {unique_count} unique values")

# Decision: Keep or Remove?
columns_to_remove = []

for col in low_variability_cols.keys():
    print(f"\nUnique values in '{col}':")
    print(df[col].unique())

    # If a column contains mostly the same value with no useful differentiation, add it to removal list
    if low_variability_cols[col] == 1:
        columns_to_remove.append(col)  # Remove if it has only one unique value

# Remove selected columns
df_filtered = df.drop(columns=columns_to_remove)

# Display the updated dataset structure
print("\nUpdated Dataset Information After Removing Low-Variability Columns:")
df_filtered.info()
```
